```python
import numpy as np
import os
import random
from PIL import Image

base_dir = 'archive/NTHU-DDD'
classes = ['drowsy', 'notdrowsy']

all_files = []
for cls in classes:
    cls_dir = os.path.join(base_dir, cls)
    for fname in os.listdir(cls_dir):
        all_files.append(os.path.join(cls_dir, fname))

random.seed(42)
random.shuffle(all_files)
sample_files = all_files[:300]

representative_frames = []   # fresh list, not the old array
for fpath in sample_files:
    img = Image.open(fpath).convert('RGB').resize((224, 224))
    arr = np.array(img, dtype=np.float32)  # raw 0-255, no /255.0
    representative_frames.append(arr)

representative_frames = np.array(representative_frames)
print(representative_frames.shape)  # (300, 224, 224, 3)
```

```python
def representative_dataset():
    for sample in representative_frames[:300]:
        sample = np.expand_dims(sample, axis=0)  # (224,224,3) -> (1,224,224,3)
        yield [sample.astype(np.float32)]
```

```python
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]

converter.representative_dataset = representative_dataset
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.int8
converter.inference_output_type = tf.int8

tflite_model = converter.convert()
with open('drowsy_model_int8.tflite', 'wb') as f:
    f.write(tflite_model)

print(len(tflite_model), "bytes")
```
